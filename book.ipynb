```python
import pandas as pd
```

```python
data = pd.read_csv("House Price India.csv")
```

```python
data.head()
```

```python
stats = data["Price"].describe().reset_index()
stats["Price"] = round(stats["Price"], 2)
```

```python
data.dropna(inplace=True)
data.drop_duplicates(inplace=True)
```

```python
data.columns
```

```python
data.head()
```

```python
import matplotlib.pyplot as plt
```

```python
data.groupby("condition of the house")["Price"].mean().plot(kind="bar")
plt.title("Condition of the house vs price")
plt.ylabel("Price")
plt.xlabel("Condition of the house")
plt.show()
```

```python
data.columns
```

```python
X = data[['number of bedrooms', 'number of bathrooms', 'living area','condition of the house','Number of schools nearby','Distance from the airport','number of floors']]
y = data[["Price"]]
```

```python
X
```

```python
y
```

```python
data.shape[0]*0.8
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2)
```

```python
X_train.shape
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeRegressor
```

```python
param_grid = {
    "criterion": ["mse", "friedman_mse", "mae"],
    "splitter": ["best", "random"],
    "max_depth": [None, 10, 20, 30,40, 50],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4], 
}
```

```python
tree_model = DecisionTreeRegressor()
```

```python
grid_tree = GridSearchCV(estimator=tree_model, param_grid=param_grid)
```

```python
grid_tree.fit(X_train, y_train)
```

```python
tree_preds=grid_tree.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error
mean_absolute_error(y_test, tree_preds)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
```

```python
lr.fit(X_train, y_train)
```

```python
predslr = lr.predict(X_test)
```

```python
mean_absolute_error(y_test, predslr)
```

```python
from sklearn.ensemble import RandomForestRegressor
rfrmodel = RandomForestRegressor()
```

```python
param_gridrfr = {
    "max_depth": [5,10,15],
    "n_estimators": [2,3,4,5,6,7,8,9,10]
}
```

```python
gridrfr = GridSearchCV(rfrmodel, param_gridrfr)
```

```python
gridrfr.fit(X_train, y_train.values.ravel())
```

```python
gridrfr.best_params_
```

```python
rfrpredictions = gridrfr.predict(X_test)
```

```python
mean_absolute_error(y_test, rfrpredictions)
```
